# How to add secrets
You *CANNOT* add secrets from inside databricks.
Instead, install the databricks CLI in powershell and execute there.

```python
%sh
#display the scopes
databricks secrets list-scopes

#enter the scope.
databricks secrets put-secret <secret scope> <key>
<value> #prompted to enter value after you enter key

```

Keys should be in CAPS_CASE

In the databricks notebook, you may now retrieve the secret

```python
jdbcHostname = dbutils.secrets.get(scope = "ops-secret-scope", key = <secret key>)
```
