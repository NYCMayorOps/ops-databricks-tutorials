```python

from shapely import wkb
import pandas as pd

# Read GeoJSON with GeoPandas
gdf_pandas = gpd.read_file("/Volumes/scorecard_fulcrum/geo/scoreacard_geospatial_files/community_districts_simplified.geojson")

print(f"Read {len(gdf_pandas)} features")
print(f"Columns: {list(gdf_pandas.columns)}")

# Convert geometry to WKB binary format
gdf_pandas['geometry'] = gdf_pandas['geometry'].apply(lambda geom: wkb.dumps(geom) if geom is not None else None)

# Convert to Spark DataFrame
gdf = spark.createDataFrame(gdf_pandas)

# Display the result
display(gdf)
```
